# Notebook 2 — Fixed-Hyperparameter Rerun

This notebook reruns selected paper points using the shipped fixed hyperparameters. Set `AEC_EXECUTE=1` before execution to start training; otherwise it produces an execution plan.

## Setup

This cell checks the environment and creates portable output directories.

```python
# Notebook configuration. Run this cell before the setup cell.
import os
AEC_EXECUTE = True  # Set True to run training; False only prints the plan.
AEC_GPU_IDS = '0,1,2,3'  # Comma-separated GPU IDs, for example '0,1'.
AEC_MAX_PARALLEL_PER_GPU = 10

if not AEC_GPU_IDS.strip(): raise ValueError('AEC_GPU_IDS cannot be empty')
if int(AEC_MAX_PARALLEL_PER_GPU) < 1: raise ValueError('AEC_MAX_PARALLEL_PER_GPU must be >= 1')
os.environ['AEC_EXECUTE'] = '1' if AEC_EXECUTE else '0'
os.environ['AEC_GPU_IDS'] = AEC_GPU_IDS
os.environ['AEC_MAX_PARALLEL_PER_GPU'] = str(AEC_MAX_PARALLEL_PER_GPU)
print({'AEC_EXECUTE': os.environ['AEC_EXECUTE'], 'AEC_GPU_IDS': os.environ['AEC_GPU_IDS'], 'AEC_MAX_PARALLEL_PER_GPU': os.environ['AEC_MAX_PARALLEL_PER_GPU']})
```

```python
from pathlib import Path
import os, sys
REPO_ROOT = Path.cwd().resolve()
while REPO_ROOT != REPO_ROOT.parent and not (REPO_ROOT / 'scripts' / 'aec').is_dir():
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT))
from scripts.aec.environment import check_environment
from scripts.aec.paths import ensure_runtime_dirs
ensure_runtime_dirs()
execute = os.environ.get('AEC_EXECUTE', '0') == '1'
ENV = check_environment(require_gpu=execute)
print('Repository:', REPO_ROOT)
```
